### NYC

```python
#NYC

len(data_nyc.query('venueCategory == "Bar"')['userId'].value_counts().unique())
```

```python
user_ci_nyc1 = data_nyc.query('venueCategory == "Bar"')
user_ci_nyc = list(user_ci_nyc1['userId'])
```

```python
from collections import Counter

user_check_in_count = Counter(user_ci_nyc)
edges = sorted([(v, k) for k, v in user_check_in_count.items()], reverse=True)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Membuat graf
G = nx.Graph()
G.add_edges_from(edges)

# Visualisasi graf
plt.figure(figsize=(20,17))
pos = nx.spring_layout(G, seed=42)  # Layout posisi node
nx.draw(G, pos, with_labels=True, node_size=100, node_color="skyblue", font_size=5, font_weight="bold")
plt.title("Graf Interaksi Sosial")
plt.show()
```

Analisis Metrik Jaringan

```python
# 1. Degree Distribution
degree_counts = [degree for node, degree in G.degree()]
print("Degree Counts (Jumlah Teman):", degree_counts)
# a. Degree Centrality
degree_centrality = nx.degree_centrality(G)
print("Degree Centrality:", degree_centrality)
# b. Betweenness Centrality
betweenness_centrality = nx.betweenness_centrality(G)
print("Betweenness Centrality:", betweenness_centrality)
# c. Closeness Centrality
closeness_centrality = nx.closeness_centrality(G)
print("Closeness Centrality:", closeness_centrality)
# 3. Community Detection (menggunakan connected components)
communities = list(nx.connected_components(G))
print("Komunitas:", communities)
```

```python
communities = list(nx.connected_components(G))
print("Komunitas:", communities)
plt.figure(figsize=(8,6))
pos = nx.spring_layout(G, seed=42)
# Warna berbeda untuk setiap komunitas
community_colors = {}
for i, community in enumerate(communities):
    for node in community:
        community_colors[node] = i

node_colors = [community_colors[node] for node in G.nodes()]
nx.draw(G, pos, with_labels=True, node_size=100, node_color=node_colors, cmap=plt.cm.get_cmap('tab20', len(communities)), font_size=1, font_weight="bold")
handles = [plt.Line2D([0], [0], marker='o', color='w', label=f"Komunitas {i+1}", markersize=10, markerfacecolor=plt.cm.get_cmap('tab20', len(communities))(i)) for i in range(len(communities))]
plt.legend(handles=handles, title="Komunitas", loc="best")
plt.title("Graf Interaksi Sosial dengan Visualisasi Komunitas Bar di NYC")
plt.savefig('nyc.png')
plt.show()
```

```python
for i, community in enumerate(communities):
  print(f"Komunitas {i+1} :{len(community)} orang")
```

visualisasi dalam bentuk spasial interaksi sosial nya

```python
userLoc = {}
for userId in user_check_in_count.keys():
    temp = user_ci_nyc1.query('userId==@userId')[['latitude', 'longitude']]
    if userId not in userLoc:
        userLoc[userId] = []
    for loc in temp.values:
        userLoc[userId].append(loc[0].astype(str) + '_' + loc[1].astype(str))

for user in userLoc.keys():
    userLoc[user] = Counter(userLoc[user])

# sorted([(k, loc, count) for k, v in userLoc.items() for loc, count in v.items()], reverse=True, key=lambda x: x[2])[:10]
spasial = sorted([(k, loc, count) for k, v in userLoc.items() for loc, count in v.items()], reverse=True, key=lambda x: x[2])
```

```python
spasial
```

```python
import folium
from collections import Counter

# Buat peta dasar (menentukan titik pusat peta berdasarkan rata-rata lat/lon)
# Misalnya titik pusatnya adalah rata-rata dari semua koordinat yang ada
center_lat = sum([float(loc.split('_')[0]) for user, loc, count in spasial]) / len(spasial)
center_lon = sum([float(loc.split('_')[1]) for user, loc, count in spasial]) / len(spasial)

# Inisialisasi peta
mymap = folium.Map(location=[center_lat, center_lon], zoom_start=12)

# Menambahkan marker pada peta untuk setiap lokasi
for user, loc, count in spasial:
    lat, lon = map(float, loc.split('_'))  # Ambil lat dan lon dari string
    folium.CircleMarker(
        location=[lat, lon],
        radius=5,
        popup=f'User: {user}, Count: {count}',  # Popup yang muncul saat mengklik marker
        color='blue',  # Warna marker
        fill=True,
        fill_color='blue'
    ).add_to(mymap)

# Simpan peta ke dalam file HTML
mymap.save('user_spatial_map.html')

# Tampilkan peta
mymap
```

```python
user_tokyo1 = data_tokyo.query('venueCategory == "Train Station"')
user_ci_tokyo = list(user_tokyo1['userId'])
```

```python
from collections import Counter

user_check_in_count = Counter(user_ci_tokyo)
edges = sorted([(v, k) for k, v in user_check_in_count.items()], reverse=True)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
# Membuat graf
G = nx.Graph()
G.add_edges_from(edges)

# 1. Degree Distribution
degree_counts = [degree for node, degree in G.degree()]
print("Degree Counts (Jumlah Teman):", degree_counts)

# Histogram Degree Distribution
plt.figure(figsize=(6,4))
plt.hist(degree_counts, bins=np.arange(1, max(degree_counts) + 2), align="left", color="orange", rwidth=0.8)
plt.title("Distribusi Jumlah Teman (Degree Distribution)")
plt.xlabel("Jumlah Teman")
plt.ylabel("Frekuensi")
plt.grid()
plt.show()

# 2. Centrality Measures
# a. Degree Centrality
degree_centrality = nx.degree_centrality(G)
print("Degree Centrality:", degree_centrality)

# b. Betweenness Centrality
betweenness_centrality = nx.betweenness_centrality(G)
print("Betweenness Centrality:", betweenness_centrality)

# c. Closeness Centrality
closeness_centrality = nx.closeness_centrality(G)
print("Closeness Centrality:", closeness_centrality)

# 3. Community Detection (menggunakan connected components)
communities = list(nx.connected_components(G))
print("Komunitas:", communities)
print("Jumlah Komunitas:", len(communities))
```

```python
communities = list(nx.connected_components(G))
print("Komunitas:", communities)
# Visualisasi komunitas
plt.figure(figsize=(12,10))
pos = nx.spring_layout(G, seed=42)
# Warna berbeda untuk setiap komunitas
community_colors = {}
for i, community in enumerate(communities):
    for node in community:
        community_colors[node] = i
node_colors = [community_colors[node] for node in G.nodes()]
nx.draw(G, pos, with_labels=True, node_size=100, node_color=node_colors, cmap=plt.cm.get_cmap('tab20', len(communities)), font_size=1, font_weight="bold")

handles = [plt.Line2D([0], [0], marker='o', color='w', label=f"Komunitas {i+1}", markersize=10, markerfacecolor=plt.cm.get_cmap('tab20', len(communities))(i)) for i in range(len(communities))]
plt.legend(handles=handles, title="Komunitas", loc="best")
plt.title("Graf Interaksi Sosial dengan Visualisasi Komunitas di Train Station")
plt.savefig('tokyo.png',dpi = 200)
plt.show()
```
